# 0.1 Import Libraries

```python
from pathlib import Path
import sys
import pandas as pd
```

# 0.2 Load Project Modules

```python
PROJECT_ROOT = Path.cwd().parents[0] if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from config.settings import PROCESSED_DIR, OUTPUT_DIR, DEFAULT_RANDOM_STATE, DEFAULT_SAMPLE_SIZE
from src.sampling import create_random_sample_panel, create_stratified_sample_panel, create_biased_large_store_panel, create_optimized_panel, compare_panel_coverage, combine_panel_store_lists
```

```python
random_panel = create_random_sample_panel(universe, DEFAULT_SAMPLE_SIZE, DEFAULT_RANDOM_STATE)
random_panel[["panel_name", "store_nbr"]].head()
```

# 2.2 Create Stratified Sample Panel

```python
stratified_panel = create_stratified_sample_panel(universe, "type", DEFAULT_SAMPLE_SIZE, DEFAULT_RANDOM_STATE)
stratified_panel[["panel_name", "store_nbr", "type"]].head()
```

# 2.4 Create Optimized Sample Panel

```python
optimized_panel = create_optimized_panel(universe, DEFAULT_SAMPLE_SIZE)
optimized_panel[["panel_name", "store_nbr", "total_sales"]].head()
```
